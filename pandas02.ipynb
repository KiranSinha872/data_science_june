```python
import pandas as pd
import numpy as np

# ============================================
# CREATE SERIES
# ============================================

print("===== SERIES =====")
s = pd.Series([10, 20, 30, 40])
print(s)

# ============================================
# CREATE DATAFRAME
# ============================================

print("\n===== DATAFRAME =====")

df = pd.DataFrame({
    "Name": ["Kiran", "Rahul", "Aman", "Priya", "Kiran"],
    "Age": [20, 25, 30, np.nan, 20],
    "Salary": [25000, 30000, 45000, 50000, 25000],
    "Department": ["IT", "HR", "IT", "Finance", "IT"],
    "Date": ["2024-01-10", "2024-02-15", "2024-03-20",
             "2024-04-25", "2024-01-10"]
})

print(df)

# ============================================
# HEAD & TAIL
# ============================================

print("\n===== HEAD =====")
print(df.head())

print("\n===== TAIL =====")
print(df.tail())

# ============================================
# INFO
# ============================================

print("\n===== INFO =====")
df.info()

# ============================================
# DESCRIBE
# ============================================

print("\n===== DESCRIBE =====")
print(df.describe())

# ============================================
# SHAPE, COLUMNS, INDEX
# ============================================

print("\nShape:", df.shape)
print("\nColumns:", df.columns)
print("\nIndex:", df.index)

# ============================================
# COLUMN SELECTION
# ============================================

print("\n===== SELECT COLUMN =====")
print(df["Name"])

# ============================================
# LOC
# ============================================

print("\n===== LOC =====")
print(df.loc[0])

print("\nName and Salary Columns:")
print(df.loc[:, ["Name", "Salary"]])

# ============================================
# ILOC
# ============================================

print("\n===== ILOC =====")
print(df.iloc[0])

print("\nFirst 3 Rows & First 2 Columns:")
print(df.iloc[0:3, 0:2])

# ============================================
# FILTERING
# ============================================

print("\n===== FILTER AGE > 25 =====")
print(df[df["Age"] > 25])

# ============================================
# ADD NEW COLUMN
# ============================================

df["Bonus"] = df["Salary"] * 0.10

print("\n===== BONUS COLUMN ADDED =====")
print(df)

# ============================================
# RENAME COLUMN
# ============================================

df.rename(columns={"Salary": "Income"}, inplace=True)

print("\n===== COLUMN RENAMED =====")
print(df)

# ============================================
# DROP COLUMN
# ============================================

df.drop("Bonus", axis=1, inplace=True)

print("\n===== BONUS DROPPED =====")
print(df)

# ============================================
# DROP NA
# ============================================

print("\n===== DROP NA =====")
print(df.dropna())

# ============================================
# SORT VALUES
# ============================================

print("\n===== SORT BY INCOME =====")
print(df.sort_values(by="Income"))

# ============================================
# SORT INDEX
# ============================================

print("\n===== SORT INDEX =====")
print(df.sort_index())

# ============================================
# ISNULL / NOTNULL
# ============================================

print("\n===== ISNULL =====")
print(df.isnull())

print("\n===== NOTNULL =====")
print(df.notnull())

# ============================================
# FILLNA
# ============================================

df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\n===== FILLNA =====")
print(df)

# ============================================
# UNIQUE
# ============================================

print("\n===== UNIQUE DEPARTMENTS =====")
print(df["Department"].unique())

# ============================================
# NUNIQUE
# ============================================

print("\n===== NUNIQUE =====")
print(df.nunique())

# ============================================
# AGGREGATE
# ============================================

print("\n===== AGG =====")
print(
    df.agg({
        "Age": ["min", "max", "mean"],
        "Income": ["sum", "mean"]
    })
)

# ============================================
# GROUPBY
# ============================================

print("\n===== GROUPBY =====")
print(df.groupby("Department")["Income"].mean())

# ============================================
# APPLY
# ============================================

print("\n===== APPLY =====")

df["Income"] = df["Income"].apply(lambda x: x + 1000)

print(df)

# ============================================
# JOIN  left right inner outer
# ============================================

print("\n===== JOIN =====")

df1 = pd.DataFrame({
    "A": [1, 2, 3]
}, index=[0, 1, 2])

df2 = pd.DataFrame({
    "B": [4, 5, 6]
}, index=[0, 1, 2])

print(df1.join(df2))

# ============================================
# MERGE
# ============================================

print("\n===== MERGE =====")

emp = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

salary = pd.DataFrame({
    "ID": [1, 2, 3],
    "Salary": [20000, 30000, 40000]
})

print(pd.merge(emp, salary, on="ID"))

# ============================================
# CONCAT
# ============================================

print("\n===== CONCAT =====")
print(pd.concat([emp, emp]))

# ============================================
# PIVOT TABLE
# ============================================

print("\n===== PIVOT TABLE =====")
print(
    df.pivot_table(
        values="Income",
        index="Department",
        aggfunc="mean"
    )
)

# ============================================
# MELT
# ============================================

print("\n===== MELT =====")
print(pd.melt(emp, id_vars=["ID"]))

# ============================================
# DATETIME
# ============================================

df["Date"] = pd.to_datetime(df["Date"])

print("\n===== YEAR =====")
print(df["Date"].dt.year)

print("\n===== MONTH =====")
print(df["Date"].dt.month)

print("\n===== DAY =====")
print(df["Date"].dt.day)

# ============================================
# STATISTICS
# ============================================

print("\nSum:")
print(df["Income"].sum())

print("\nMean:")
print(df["Income"].mean())

print("\nMedian:")
print(df["Income"].median())

print("\nMin:")
print(df["Income"].min())

print("\nMax:")
print(df["Income"].max())

print("\nCount:")
print(df.count())

print("\nStd:")
print(df["Income"].std())

print("\nVariance:")
print(df["Income"].var())

# ============================================
# CORRELATION
# ============================================

print("\n===== CORRELATION =====")
print(df[["Age", "Income"]].corr())

# ============================================
# CUMULATIVE FUNCTIONS
# ============================================

print("\n===== CUMSUM =====")
print(df["Income"].cumsum())

print("\n===== CUMMAX =====")
print(df["Income"].cummax())

print("\n===== CUMMIN =====")
print(df["Income"].cummin())

# ============================================
# DUPLICATES
# ============================================

print("\n===== DUPLICATED =====") #true or false
print(df.duplicated())

print("\n===== DROP DUPLICATES =====")
print(df.drop_duplicates())

# ============================================
# SAMPLE
# ============================================

print("\n===== SAMPLE =====")
print(df.sample(2)) # returns two random rows,select random rows from a DataFrame. for testing ananylises

# ============================================
# TRANSPOSE
# ============================================

print("\n===== TRANSPOSE =====")
print(df.T)

# ============================================
# NUMPY ARRAY
# ============================================

print("\n===== TO NUMPY =====")
print(df.to_numpy())

# ============================================
# DICTIONARY
# ============================================

print("\n===== TO DICT =====")
print(df.to_dict())

# ============================================
# SAVE FILES
# ============================================

df.to_csv("output.csv", index=False)
df.to_excel("output.xlsx", index=False)
df.to_json("output.json")

print("\nFiles Saved Successfully!")
```
