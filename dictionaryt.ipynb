```python
#DICTIONARY

#A dictionary stores data in key : value pairs.

dhruv={
    "name":"dhurv tomar",
    "back":"maths",
    "year":1,
    "city": "Ghaziabad"

}
print(dhruv) #{'name': 'dhurv tomar', 'back': 'maths', 'year': 1, 'city': 'Ghaziabad'}



#accesing a key
print(dhruv["city"])
print(dhruv["year"])

#adding a new key
dhruv["sex"]="male"
print(dhruv)


#updating a key
dhruv["year"]=2
print(dhruv)

#removing a key
# use pop(key)
dhruv.pop("back")
print(dhruv)






```
