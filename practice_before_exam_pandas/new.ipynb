```python
import matplotlib.pyplot as plt

plt.scatter(df['study_hours'], df['exam_score'])
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.show()
```

```python
plt.scatter(df['attendance'], df['exam_score'])
plt.xlabel('Attendance')
plt.ylabel('Exam Score')
plt.title('Attendance vs Exam Score')
plt.show()
```

```python
status_count = df['placement_status'].value_counts()

plt.bar(status_count.index, status_count.values)
plt.xlabel('Placement Status')
plt.ylabel('Count')
plt.title('Placement Distribution')
plt.show()
```

```python
plt.hist(df['exam_score'], bins=20)
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.title('Exam Score Distribution')
plt.show()
```

```python
avg_score = df.groupby('study_hours')['exam_score'].mean()

plt.bar(avg_score.index, avg_score.values)
plt.xlabel('Study Hours')
plt.ylabel('Average Exam Score')
plt.title('Average Score by Study Hours')
plt.show()
```
