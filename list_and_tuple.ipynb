```python
#LIST
#muttable
l=[1,2,3,4,4,"kiran"]
print(l) #[1, 2, 3, 4, 4, 'kiran']


l.append(5)
print(l) #[1, 2, 3, 4, 4, 'kiran', 5]

#list in list
l.append([8,7,6]) #[1, 2, 3, 4, 4, 'kiran', 5, [8, 7, 6]]
print(l)

#extend
l.extend([8,7,6])
print(l) #[1, 2, 3, 4, 4, 'kiran', 5, [8, 7, 6], 8, 7, 6]


#insert(indx,ele)
#to insert element anywhere
l.insert(3,100)
l.insert(4,"kirandaddy")
print(l)

#pop()
#remove last element
l.pop()
# l.pop(3) # remove element at index 3
print(l)


#remove(ele)
#removes the element from first occurrence
l.remove(4)
print(l)


#index(ele)
#gives the index of the element
i=l.index(100)
print(i)


#count(ele)
x=l.count(2)
print(x)


#sort()
#it uses Timsort
#sort() returns None, so we have to use it separately and then print the list
a=[1,4,2,5,6,7,2,3,4,9,1,56]
a.sort()
print(a)

#sorted()
#sorted() returns a new list. Original list remains unchanged.
b=sorted(a)
print(b)


#sum(list),max(list),min()
```

```python
#sum of two list

n = int(input("Enter size of lists: "))

list1 = []
list2 = []

print("Enter elements of first list:")
for i in range(n):
    list1.append(int(input()))

print("Enter elements of second list:")
for i in range(n):
    list2.append(int(input()))

ans = []

for i in range(n):
    ans.append(list1[i] + list2[i])

print("Sum list:", ans)



```

```python
#TUPLES
#immutable

tp=(1,2,5,4)
print(tp) #(1, 2, 3, 4)

# tp.sort() error because tuples are immutable
#make a new tuple

# a=sorted(tp) #[1, 2, 3, 4] sorted return a list
a=tuple(sorted(tp))# (1, 2, 3, 4)
print(a)



p=(1,3,5,7,9,8,6,4,2,7,9,8,6,4,2)

#count(ele)
# to count elements
print(p.count(2))

#index()
#to find the index of the element
print(p.index(2))


#len(tuple name)
print(len(p))

#sum(tuple name)
print(sum(p))

print(max(p))
print(min(p))

```

```python
# function in python
# def means define a function

def greet():
    print("hiii")

# calling
greet()


# function with parameter
def her(name):
    print("hiii", name)

# calling
her("kiran")


# function that returns a value
def addd(x, y):
    return x + y

result = addd(1, 2)
print(result)
```
