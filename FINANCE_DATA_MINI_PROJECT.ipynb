```python
# dp.groupby('loan_status')['id'].count()
loan_status_data = dp.groupby('loan_status').agg({
    'id': 'count',
    'dti': 'mean',
    'int_rate': 'mean',
    'loan_amount': 'sum',
    'loan_amount': 'sum',
    'total_payment': 'sum'
})

loan_status_data.rename(
    columns={'id': 'number of applications'},
    inplace=True
)

display(loan_status_data)
```

```python
#good loan


#no of application
no_of_app=loan_status_data.loc['Current','number of applications']+loan_status_data.loc['Fully Paid','number of applications']
print(no_of_app)

#funded amount
funded_amount=loan_status_data.loc['Current','loan_amount']+loan_status_data.loc['Fully Paid','loan_amount']
print(funded_amount)

#recievded amount
recieved_amount=loan_status_data.loc['Current','total_payment']+loan_status_data.loc['Fully Paid','total_payment']
print(recieved_amount)

#profit
print(recieved_amount-funded_amount)
```

```python
#bad loan


# Number of bad loan applications
bad_loan_apps = loan_status_data.loc['Charged Off', 'number of applications']
print(bad_loan_apps)

# Funded amount for bad loans
bad_funded_amount = loan_status_data.loc['Charged Off', 'loan_amount']
print(bad_funded_amount)

# Amount received from bad loans
bad_received_amount = loan_status_data.loc['Charged Off', 'total_payment']
print(bad_received_amount)

# Loss on bad loans
print(bad_received_amount - bad_funded_amount)
```
